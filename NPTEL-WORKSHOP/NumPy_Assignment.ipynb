## Question 12  ·  *2 marks*
### Stacking and splitting

Let `P = np.arange(6).reshape(2, 3)` and `Q = np.arange(6, 12).reshape(2, 3)`.

**(a)** Join them side by side and print the shape of the result.
**(b)** Join them one above the other and print the shape.
**(c)** Split the horizontally-joined array back into **2 equal parts**.

````python
P = np.arange(6).reshape(2, 3)
Q = np.arange(6, 12).reshape(2, 3)

# (a) Side by side (horizontal)
h_joined = np.hstack((P, Q))
print("(a) Horizontal shape:", h_joined.shape)

# (b) One above the other (vertical)
v_joined = np.vstack((P, Q))
print("(b) Vertical shape:", v_joined.shape)

# (c) Split horizontally-joined array into 2 equal parts
left, right = np.hsplit(h_joined, 2)
print("\n(c) Split Part 1:\n", left)
print("    Split Part 2:\n", right)
````

## Question 16  ·  *3 marks*
### Write a formula

Write a function `normalise(a)` that scales any array to the range $[0, 1]$ using

$$x_{\text{norm}} = \frac{x - x_{\min}}{x_{\max} - x_{\min}}$$

Apply it to `arr` and print the result rounded to 2 decimals, along with its **minimum** and **maximum**
(which must be 0 and 1).

````python
def normalise(a):
    return (a - a.min()) / (a.max() - a.min())

norm_arr = normalise(arr)
print("Normalised arr (rounded to 2 decimals):\n", np.round(norm_arr, 2))
print("Minimum:", norm_arr.min())
print("Maximum:", norm_arr.max())
````

**(a)** Create the array $[10, 20, \text{nan}, 40, \text{nan}, 60]$ and print its mean —
explain the result in a comment.

**(c)** Plot $y = \sin(x)$ for 100 points between $-2\pi$ and $2\pi$, with a title and a grid.

## Question 20  ·  *4 marks*
### Mini case study — exam analysis

The array below holds the marks of **6 students** (rows) in **4 subjects** (columns).

```python
scores = np.array([[72, 65, 80, 90],
                   [55, 40, 62, 58],
                   [88, 92, 79, 95],
                   [30, 45, 50, 38],
                   [67, 71, 60, 74],
                   [95, 89, 91, 87]])
```

Answer all five parts:

**(a)** Print each student's **total** and **average** mark.
**(b)** Print the **subject-wise topper's marks** (the highest mark in each subject).
**(c)** Print the **index** of the student with the highest total.
**(d)** Print the marks of only those students whose **average is above 70**.
**(e)** Create a pass/fail array: `1` where a mark is $\ge 50$ and `0` otherwise, and print the
**number of passes in each subject**.

````python
scores = np.array([[72, 65, 80, 90],
                   [55, 40, 62, 58],
                   [88, 92, 79, 95],
                   [30, 45, 50, 38],
                   [67, 71, 60, 74],
                   [95, 89, 91, 87]])

# (a) Each student's total and average mark
student_totals = scores.sum(axis=1)
student_avgs = scores.mean(axis=1)
print("(a) Student Totals:", student_totals)
print("    Student Averages:", student_avgs)

# (b) Subject-wise topper's marks (highest mark in each subject)
subject_toppers = scores.max(axis=0)
print("\n(b) Subject-wise Topper Marks:", subject_toppers)

# (c) Index of the student with the highest total
top_student_idx = student_totals.argmax()
print("\n(c) Index of Student with Highest Total:", top_student_idx)

# (d) Marks of only those students whose average is above 70
above_70_marks = scores[student_avgs > 70]
print("\n(d) Marks of Students with Average > 70:\n", above_70_marks)

# (e) Pass/fail array (1 if >= 50 else 0) and number of passes in each subject
pass_fail = np.where(scores >= 50, 1, 0)
passes_per_subject = pass_fail.sum(axis=0)
print("\n(e) Pass/Fail Array:\n", pass_fail)
print("    Number of Passes in Each Subject:", passes_per_subject)
````
