A **vector** (1-D array) with 3 elements:

$$v = \begin{bmatrix} 1 & 2 & 3 \end{bmatrix}$$

A **matrix** (2-D array) with 2 rows and 3 columns — written $A_{2\times 3}$:

$$A = \begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{bmatrix}$$

A **3-D array** is a *stack of matrices*:

$$T[0] = \begin{bmatrix} 0 & 1 \\ 2 & 3 \end{bmatrix}, \qquad
  T[1] = \begin{bmatrix} 4 & 5 \\ 6 & 7 \end{bmatrix}$$

The **shape** is written $(\text{rows},\ \text{columns})$ — so $A$ has shape $(2, 3)$ and $T$ has shape $(2, 2, 2)$.

## 1. What is NumPy?

**NumPy** stands for **Numerical Python**.

> NumPy is a library used for working with **arrays** — and it is the **fundamental package for
> scientific computing** in Python.

**In simple words:** NumPy lets you store numbers in a fast container and do maths on *all* of them at once.

#### The core object — the `ndarray`

An `ndarray` is an **n-dimensional array**: a grid of values, **all of the same type**, indexed by a
tuple of non-negative integers.

| Dimensions | Maths name | Example |
|---|---|---|
| 1-D | Vector | $\begin{bmatrix} 1 & 2 & 3\end{bmatrix}$ |
| 2-D | Matrix | $\begin{bmatrix} 1 & 2 \\ 3 & 4\end{bmatrix}$ |
| 3-D and above | Tensor | a stack of matrices |

---

## 2. Why do we need NumPy?

Python already has lists — so why another container?

1. **Speed.** Python is an interpreted language and its lists are slow for number crunching.
   NumPy's core is written in **C**, so the loop runs at C speed.
2. **Memory.** A Python list stores *pointers* to objects scattered in memory. A NumPy array stores the
   numbers themselves in **one continuous block**.
3. **Convenience.** `a + b` adds two whole arrays — no `for` loop, no `append`.

**Built on NumPy:** Pandas · Matplotlib · Scikit-learn · TensorFlow · PyTorch · OpenCV.
Learning NumPy first makes every one of these easier.

---

## 3. NumPy arrays vs Python sequences

| | Python list | NumPy array |
|---|---|---|
| Item types | Can be mixed | All the same (`dtype`) |
| Size | Can grow and shrink | **Fixed** at creation |
| Maths | Needs a loop | Applied to the whole array |
| Memory | Scattered pointers | One compact block |
| Speed on large data | Slow | Fast (C underneath) |

## 6. Changing the data type — `astype()`

Real datasets are large, and memory matters. If your numbers are small whole numbers, storing them as
`int64` (8 bytes) wastes memory — `int32` (4 bytes) or even `int8` is enough.

| dtype | Bytes | Range |
|---|---|---|
| `int8` | 1 | $-128 \ldots 127$ |
| `int32` | 4 | about $\pm 2.1 \times 10^{9}$ |
| `int64` | 8 | very large |

⚠️ Choose a type your values actually fit into — otherwise the numbers overflow and go wrong.

**👉 In the next cell:** we convert a float array to `int32` with `astype`. Note two things: the decimals are **cut off**
(not rounded), and `astype` returns a **new** array.

````python
f = np.array([1.7, 2.3, 3.9, 4.5])
f.dtype
````

````python
g = f.astype(np.int32)
g
````

## 14. NumPy array vs Python list — a practical comparison

### 14.1 Speed

We add **one crore (10 million)** numbers, first with a Python list, then with a NumPy array:

$$c_i = a_i + b_i, \qquad i = 0, 1, \ldots, 10^{7}-1$$

**👉 In the next cell:** the Python version: two lists and a list comprehension, timed with `time.time()`.
This takes a couple of seconds — that wait is the point of the demo.

````python
import time

a = list(range(10_000_000))
b = list(range(10_000_000, 20_000_000))

start = time.time()
c = [a[i] + b[i] for i in range(len(a))]
python_time = time.time() - start
python_time
````

**👉 In the next cell:** the same calculation in NumPy — one line, no loop — and the speed-up factor.

````python
a = np.arange(10_000_000)
b = np.arange(10_000_000, 20_000_000)

start = time.time()
c = a + b
Numpy_time = time.time() - start
Numpy_time
````

### 14.2 Memory

**👉 In the next cell:** the memory used by the same 10 million numbers. `sys.getsizeof` measures the Python list container,
`nbytes` measures the array exactly. The array wins by a wide margin — and the gap is even larger than
it looks, because each Python integer is a separate object of about 28 bytes.

### 14.3 Convenience

Python list: `[a[i] + b[i] for i in range(len(a))]`
NumPy array: `a + b`

Speed, memory **and** readability — all three.

### 15.2 Boolean indexing ⭐

The most used feature in all of data science — and Pandas filtering works exactly the same way.

Formally we are keeping the set

In code it is **two steps**:

## 16. Broadcasting

**Definition.** Broadcasting describes how NumPy treats arrays of **different shapes** during arithmetic.
The smaller array is *stretched* across the larger one — **without copying any data**.

You have used it already: `a * 2` stretches the single number 2 across the whole array.

**👉 In the next cell:** a $(3,4)$ array plus a $(4,)$ vector. The vector is added to **every row**, because its length matches
the number of columns.

````python
a = np.arange(12).reshape(3,4)
b = np.arange(4)
````

````python
a
````

````python
b
````

````python
a + b
````

### 16.1 The broadcasting rules

Compare the shapes **from the right**:

**Rule 1 — pad on the left.** If one array has fewer dimensions, put $1$s on the left of its shape:
$(4,) \rightarrow (1, 4)$.

**Rule 2 — equal or one.** In each dimension the sizes must be **equal**, or **one of them must be 1**
(the 1 is then stretched).

If neither holds → `ValueError`.

$$\underbrace{(3,4)}_{a} + \underbrace{(4,)\rightarrow(1,4)}_{b} \;\Rightarrow\; (3,4) \;\;✅$$
$$\underbrace{(3,4)}_{a} + \underbrace{(3,)\rightarrow(1,3)}_{b} \;\Rightarrow\; 4 \neq 3 \;\;❌$$

## 18. Working with missing values

Real data has gaps. NumPy writes a missing number as **`np.nan`** (*Not a Number*), and **any** arithmetic
touching it returns `nan`.

**👉 In the next cell:** an array with one missing value, and its mean — which comes out as `nan`.

````python
a = np.array([1, 2, 3, 4, np.nan])
a
````

**👉 In the next cell:** the fix: `np.isnan(a)` marks the missing entries, and `~` flips the mask so we keep the rest.
The mean is then computed on clean data only.

````python
a[~np.isnan(a)]
````
