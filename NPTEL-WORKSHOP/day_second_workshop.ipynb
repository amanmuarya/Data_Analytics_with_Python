```python
import pandas as pd 
import numpy as np
```

```python
np.random.seed(42) # same random numbers for everyone
# temps: Delhi's daily maximum temperature for one year (simulated)
monthly_avg = [21, 24, 30, 36, 40, 39, 35, 34, 34, 33, 28, 23] # approx. average max °C, Jan → Dec
mid_month = [15, 45, 74, 105, 135, 166, 196, 227, 258, 288, 319, 349] # day number of each month's middle
day = np.arange(365) # 0, 1, 2, …, 364
seasonal = np.interp(day, mid_month, monthly_avg, period = 365) # smooth yearly curve
temps = pd.Series(
    (seasonal + np.random.normal(0, 1.8, 365)).round(1),
    name = 'max_temp_c'
) # kohli: runs in each IPL innings (simulated)
kohli = pd.Series(
    (np.random.beta(0.9, 2.2, 215) * 120).round().astype(int),
    name = 'runs'
) 
print(temps.size, kohli.size)
```
