here find min and max

```python
data = [23, 34, 45, 56, 11, 33, 89, 65, 33, 22 , 90 ]
def min_and_max(data):
    min_val = min(data)
    max_val = max(data)

    return (min_val , max_val)             # use function 
```

```python
min_and_max(data)
```

here use function 

```python
def ran_get(data):
    min_val = min(data)
    max_val = max(data)

    return (min_val - max_val)  
```

```python
ran_get(data)
```
